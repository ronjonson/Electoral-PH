```python
import os

import pandas as pd
```

```python
DIR = r'C:\Users\Neil\Documents\Projects\Electoral-PH\Electoral-PH\image_creator\static\data'
```

```python
files = os.listdir(DIR)[:-4]
election_data = []
for file in files:
    print(file[:4])
    if 'Reps' not in file:
        election_data.append(file)
        if 'Vice' in file:
            print(f'VICE PRES: {file}')
        else:
            print(f'PRES: {file}')
```

```python
dfs = []
for file in election_data:
    df = pd.read_csv(DIR+'/'+file, index_col=0)
    cols = df.columns[1:]
    #df.melt(id_vars=['province_id'], value_vars=['Gloria_Arroyo', 'Fernando_Poe_Jr', 'Panfilo_Lacson', 'Raul_Roco', 'Eddie_Villanueva'])
    df = df.melt(id_vars=['province_id'], value_vars=cols)

    df['year'] = file[:4]
    df['position'] = "Vice_President" if 'Vice' in file else "President"
    df.rename(columns={'variable': 'candidate', 'value':'votes'}, inplace=True)
    
    dfs.append(df)
```

```python
master_df = pd.concat(dfs)
```

```python
master_df = master_df[['year', 'position', 'province_id', 'candidate','votes']]
```
